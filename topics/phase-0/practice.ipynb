```python
import tiktoken
enc = tiktoken.encoding_for_model("gpt-4o")

text = "Hello, my name ain't Sahil and I am an AI engineer."
tokens = enc.encode(text)
print(len(tokens))  # → 13 tokens
print(enc.decode(tokens))


# GPT-4o pricing: $5 per 1M input tokens
# 12 tokens = $0.00006 — tiny, but at scale it adds up fast
```

```python
print("\nTokens as strings:")
for token in tokens:
    print(repr(enc.decode_single_token_bytes(token).decode("utf-8", errors="replace")))
```

```python
for t in tokens:
    s = enc.decode_single_token_bytes(t).decode("utf-8", errors="replace")
    print(f"{t} -> {repr(s)}")
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np

# Load local embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')

def cosine_similarity(a, b):
    return np.dot(a, b) / (
        np.linalg.norm(a) * np.linalg.norm(b)
    )

texts = [
    "What is the capital of France?",
    "Paris is the capital city of France.",
    "I love eating pizza."
]

# Generate embeddings locally
embeddings = model.encode(texts)

embed_1 = embeddings[0]
embed_2 = embeddings[1]
embed_3 = embeddings[2]

sim_1_2 = cosine_similarity(embed_1, embed_2)
sim_1_3 = cosine_similarity(embed_1, embed_3)

print("1 vs 2:", sim_1_2) # 84%
print("1 vs 3:", sim_1_3) # 11%
```
